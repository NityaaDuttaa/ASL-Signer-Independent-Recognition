```python

import numpy as np

features_path = os.path.join(
    project_dir,
    "verified_reconstructed_features.npz"
)

with np.load(features_path, allow_pickle=True) as data:
    X = data["X"]
    splits = data["split"]

# Select one validation sample
val_indices = np.where(splits == "val")[0]

if len(val_indices) == 0:
    raise ValueError("No validation samples found.")

sample_index = val_indices[0]
sample = X[sample_index].astype(np.float32)

# Apply the saved training scaler
with np.load(scaler_path) as scaler:
    mean = scaler["mean"]
    scale = scaler["scale"]

scale = np.where(scale == 0, 1.0, scale)
sample_scaled = (sample - mean) / scale

# Predict the sign
probabilities = model.predict(
    sample_scaled[np.newaxis, :, :],
    verbose=0
)[0]

top_indices = np.argsort(probabilities)[-5:][::-1]

print("Input shape:", sample.shape)
print("\nTop 5 predicted signs:")

for index in top_indices:
    print(f"{classes[index]}: {probabilities[index]:.4f}")
```

```python

import gradio as gr
import numpy as np

def predict_asl(features_text):
    try:
        values = np.fromstring(features_text.replace(",", " "), sep=" ")

        if values.size != 8 * 258:
            return {
                "Error": (
                    f"Expected 2064 numbers (8 frames × 258 features), "
                    f"but received {values.size}."
                )
            }

        sequence = values.reshape(8, 258).astype(np.float32)

        with np.load(scaler_path) as scaler:
            mean = scaler["mean"]
            scale = scaler["scale"]

        scale = np.where(scale == 0, 1.0, scale)
        sequence = (sequence - mean) / scale

        probabilities = model.predict(
            sequence[np.newaxis, :, :],
            verbose=0
        )[0]

        top_indices = np.argsort(probabilities)[-5:][::-1]

        return {
            str(classes[i]): float(probabilities[i])
            for i in top_indices
        }

    except Exception as e:
        return {"Error": str(e)}


demo = gr.Interface(
    fn=predict_asl,
    inputs=gr.Textbox(
        lines=5,
        label="Landmark features",
        placeholder="Enter 2064 numerical values separated by spaces or commas"
    ),
    outputs=gr.Label(num_top_classes=5, label="Top 5 predictions"),
    title="ASL Sign Recognition",
    description=(
        "Predicts one of 100 isolated ASL gloss classes from "
        "8 frames of pre-extracted landmark features."
    )
)

demo.launch(share=True)
```

```python

import os
import json

parts_dir = os.path.join(
    project_dir, "recovered_notebook_parts"
)

keywords = [
    "PoseLandmarker",
    "HandLandmarker",
    "pose_landmarker.task",
    "hand_landmarker.task",
    "VideoCapture",
    "landmark",
    "extract",
]

for filename in sorted(os.listdir(parts_dir)):
    if not filename.endswith(".ipynb"):
        continue

    path = os.path.join(parts_dir, filename)

    with open(path, "r", encoding="utf-8") as f:
        notebook = json.load(f)

    matches = []

    for cell_number, cell in enumerate(notebook.get("cells", [])):
        if cell.get("cell_type") != "code":
            continue

        source = "".join(cell.get("source", []))

        if any(word.lower() in source.lower() for word in keywords):
            matches.append((cell_number, source))

    print(f"\n{'=' * 20} {filename} {'=' * 20}")

    if not matches:
        print("No matching code found.")
    else:
        for cell_number, source in matches[:5]:
            print(f"\nCell {cell_number}:\n{source[:1800]}")
```

```python

from pathlib import Path

PROJECT_DIR = Path("/content/drive/MyDrive/ASL_Citizen_Project")
app_path = PROJECT_DIR / "app.py"

app_code = r'''
import json
import numpy as np
import gradio as gr
import tensorflow as tf
from pathlib import Path

BASE = Path(__file__).parent
MODEL_DIR = BASE / "final_model"

model = tf.keras.models.load_model(
    MODEL_DIR / "asl_signer_independent_lstm.keras"
)

labels = np.load(MODEL_DIR / "label_classes.npy", allow_pickle=False)

with np.load(MODEL_DIR / "preprocessing_scaler.npz") as data:
    scaler_mean = data["mean"].astype(np.float32)
    scaler_scale = data["scale"].astype(np.float32)

def predict_features(text):
    try:
        values = np.fromstring(text.replace(",", " "), sep=" ")

        if values.size != 8 * 258:
            return {
                "Error": (
                    f"Expected 2064 numbers (8 frames × 258 features); "
                    f"received {values.size}."
                )
            }

        if not np.isfinite(values).all():
            return {"Error": "Input contains NaN or infinite values."}

        X = values.reshape(1, 8, 258).astype(np.float32)
        X = (X - scaler_mean) / np.maximum(scaler_scale, 1e-7)

        probabilities = model.predict(X, verbose=0)[0]
        top_indices = np.argsort(probabilities)[-5:][::-1]

        return {
            str(labels[i]): float(probabilities[i])
            for i in top_indices
        }

    except Exception as exc:
        return {"Error": str(exc)}

with gr.Blocks(title="ASL Sign Classifier") as demo:
    gr.Markdown(
        "# ASL Sign Classifier\n"
        "Classifies one pre-extracted sequence of 8 frames and 258 features "
        "into one of 100 trained sign classes.\n\n"
        "**This demo does not accept raw videos or webcam input.**"
    )

    features = gr.Textbox(
        label="Feature values",
        placeholder="Enter 2064 numbers separated by spaces or commas",
        lines=5
    )
    output = gr.Label(label="Top 5 predictions", num_top_classes=5)
    button = gr.Button("Predict")
    button.click(predict_features, inputs=features, outputs=output)

if __name__ == "__main__":
    demo.launch()
'''

app_path.write_text(app_code, encoding="utf-8")
print("Created:", app_path)
print("File size:", app_path.stat().st_size, "bytes")
```
