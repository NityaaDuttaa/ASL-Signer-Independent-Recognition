```python
import os
import json
import numpy as np

project_dir = "/content/drive/MyDrive/ASL_Citizen_Project"
batch_dir = os.path.join(project_dir, "batches")

batch_files = sorted(
    f for f in os.listdir(batch_dir) if f.endswith(".npz")
)

for filename in [batch_files[0], batch_files[len(batch_files)//2], batch_files[-1]]:
    path = os.path.join(batch_dir, filename)

    with np.load(path, allow_pickle=False) as data:
        print("\nBATCH:", filename)
        print("Feature shape:", data["features"].shape)
        print("Metadata examples:")

        for item in data["metadata"][:3]:
            print(item)

        features = data["features"]
        print("Finite values:", np.isfinite(features).all())
        print("Nonzero fraction:", np.count_nonzero(features) / features.size)

with np.load(os.path.join(project_dir, "verified_reconstructed_features.npz"), allow_pickle=False) as data:
    X = data["X"]

print("\nFEATURE POSITION ANALYSIS")
print("Shape:", X.shape)

for start, end, name in [
    (0, 132, "First 132 features"),
    (132, 195, "Next 63 features"),
    (195, 258, "Last 63 features")
]:
    group = X[:, :, start:end]
    print(
        name,
        "| shape:", group.shape,
        "| mean:", round(float(group.mean()), 4),
        "| std:", round(float(group.std()), 4),
        "| nonzero fraction:", round(float(np.count_nonzero(group) / group.size), 4)
    )
```
